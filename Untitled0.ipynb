```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score

# Main objective: Classify emails as spam or legitimate
# Method: Truncated SVD -> WKNN (Weighted K-Nearest Neighbors)
```

## 1. Load Data
You'll need to load your email dataset here. Ensure it has a column for text and a column for labels (e.g., 'spam', 'legitimate').

**Example:**
`data = pd.read_csv('your_email_dataset.csv')`
`X = data['text']  # Email content`
`y = data['label'] # Spam/legitimate labels`

For this example, we'll use a small, in-memory dataset.

```python
X = [
    "Free money now! Click here!",
    "Hello, how are you doing today?",
    "Urgent: Your account is suspended, verify now!",
    "Meeting scheduled for tomorrow at 10 AM.",
    "Congratulations! You've won a prize!",
    "Regarding the project deadline extension."
]
y = ["spam", "legitimate", "spam", "legitimate", "spam", "legitimate"]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Training samples: {len(X_train)}")
print(f"Testing samples: {len(X_test)}")
```

## 2. Text Preprocessing - TF-IDF Vectorization
Convert text data into numerical features using TF-IDF (Term Frequency-Inverse Document Frequency).

```python
vectorizer = TfidfVectorizer(stop_words='english', max_features=2000) # Limit features to avoid curse of dimensionality
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print(f"Original TF-IDF dimensions (train): {X_train_tfidf.shape}")
```

## 3. Dimensionality Reduction - Truncated SVD
Apply Truncated SVD to reduce the dimensionality of the TF-IDF features. This helps in reducing noise and computational cost. Choose the number of components based on your dataset and desired information retention.

```python
n_components = min(50, X_train_tfidf.shape[1] - 1) # Ensure n_components is less than number of features
if n_components < 1: # Handle cases where there might be very few features
    n_components = 1

svd = TruncatedSVD(n_components=n_components, random_state=42)
X_train_svd = svd.fit_transform(X_train_tfidf)
X_test_svd = svd.transform(X_test_tfidf)

print(f"Dimensions after Truncated SVD (train): {X_train_svd.shape}")
```

## 4. Classification - WKNN (Weighted K-Nearest Neighbors)
Use KNeighborsClassifier. Setting 'weights' to 'distance' implements WKNN, where closer neighbors have a greater influence on the classification.

```python
k = 3 # You can tune this parameter (number of neighbors)
wknn_model = KNeighborsClassifier(n_neighbors=k, weights='distance', n_jobs=-1) # n_jobs=-1 uses all available processors
wknn_model.fit(X_train_svd, y_train)
```

## 5. Evaluate the Model

```python
y_pred = wknn_model.predict(X_test_svd)

print("\n--- Model Evaluation ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
